# EXPLICACIÓN DEL TEMA E HIPÓTESIS

El objetivo social y financiero consiste en incrementar el capital de las personas en el muy corto plazo (ventanas móviles de 7 a 30 días) con el propósito de generar riqueza y mitigar la vulnerabilidad económica.

# Planteamiento e hipótesis

### Planteamiento de Hipótesis (Pre-Análisis)

**Hipótesis Inicial ($H_0$ - Creencia popular):** 

Para obtener grandes beneficios a muy corto plazo e incrementar rápidamente el capital de las personas, se debe invertir el 100 % del capital en los activos con mayor rendimiento nominal histórico reciente (criptoactivos como BTC-USD o renta variable tecnológica agresiva como QQQ).

**Hipótesis Alternativa ($H_1$ - Descubrimiento empírico esperado):** 

Invertir exclusivamente en activos de alta rentabilidad nominal a muy corto plazo expone a las personas a un Riesgo de Cola (Valor en Riesgo, VaR) y a caídas máximas (Maximum Drawdown) inasumibles para economías vulnerables. La mayor eficiencia en el crecimiento del capital a corto plazo (medida por el Ratio de Sharpe y el Ratio Sortino) se logra combinando un núcleo de preservación de capital (Renta Fija a corto plazo / Oro) con una exposición controlada al impulso (momentum) de renta variable.

## Aclaraciones

### VENTANAS MÓVILES

Las ventanas móviles (o rolling windows) son una técnica estadística que divide un historial de datos financieros en bloques consecutivos y superpuestos de duración fija para analizar cómo evoluciona un activo o cartera a lo largo del tiempo.
En lugar de mirar un único dato final desde el inicio hasta hoy, este método desliza un marco temporal (por ejemplo, de 12 meses o 3 años) día a día o mes a mes sobre toda la serie histórica.

**Cómo funcionan en la práctica**

*   Desplazamiento continuo: Si analizas una ventana móvil de 3 años sobre un historial de 10 años, el primer bloque irá de enero de 2016 a diciembre de 2018. El siguiente bloque avanzará un mes (de febrero de 2016 a enero de 2019), y así sucesivamente hasta el presente.
*   Aplicación a indicadores: Se utilizan para calcular rentabilidades anualizadas, volatilidad, correlaciones o métricas como los ratios de Sharpe y Sortino de forma dinámica.

**¿Por qué son útiles para el inversor?**

*   Eliminan el sesgo de la fecha de inicio. La rentabilidad acumulada tradicional depende mucho del día exacto en que empezaste a invertir. Las ventanas móviles muestran qué habría ocurrido si hubieras entrado en cientos de momentos distintos.
*   Evalúan la consistencia. Permiten comprobar si un fondo de inversión mantiene buenos resultados de forma estable o si su éxito se debe únicamente a un golpe de suerte en un año concreto.
*   Miden el riesgo real por plazos. Ayudan a visualizar cuál fue el peor escenario posible (drawdown o caída máxima) y el mejor escenario dentro de un horizonte temporal específico, como periodos de 5 años.

### RIESGO DE COLA

El riesgo de cola (o tail risk) es la probabilidad de que una inversión sufra una pérdida extrema e inesperada, mucho mayor de lo que predicen los modelos financieros tradicionales.
El nombre proviene de la estadística. Si representas los rendimientos de un activo en una gráfica de distribución normal (campana de Gauss), la gran mayoría de los movimientos diarios se agrupan en el centro. Los extremos izquierdo y derecho se estrechan formando dos "colas": 
* la cola izquierda representa las caídas catastróficas
* la cola derecha representa las ganancias extraordinarias e inesperadas, es decir, subidas de precio mucho mayores de lo habitual.

En una distribución de rendimientos, los dos extremos reflejan los escenarios poco frecuentes, pero en direcciones opuestas:
* Subidas explosivas: Incluye eventos como el descubrimiento de un fármaco revolucionario por una biotecnológica, la compra de una empresa con una prima muy alta o un rebote histórico del mercado tras una crisis.
* Riesgo de cola derecha (para posiciones cortas): Aunque para un inversor tradicional la cola derecha es una buena noticia, supone una pérdida catastrófica para quienes apuestan a la baja (short sellers) o venden opciones sin cobertura, ya que sus pérdidas pueden ser ilimitadas si el activo se dispara.
* Relación con el ratio de Sortino: Como el ratio de Sortino ignora la volatilidad positiva, no penaliza a los activos que tienen una cola derecha pronunciada. De hecho, los fondos de capital riesgo (venture capital) o las estrategias de seguimiento de tendencias (trend following) buscan precisamente capturar este extremo para compensar múltiples pérdidas pequeñas.

**En qué consiste y por qué ocurre**

* Fallo de la distribución normal: Los modelos clásicos asumen que las variaciones extremas (a más de tres desviaciones estándar de la media) son casi imposibles. En la realidad del mercado, estas caídas ocurren con mucha más frecuencia de lo que indica la teoría ("colas gruesas" o fat tails).
* Eventos de alto impacto: Se desencadena por crisis sistémicas, quiebras inesperadas, conflictos geopolíticos, pandemias o fallos tecnológicos en los mercados (como el desplome de 2008 o la caída por el COVID-19 en marzo de 2020).
* Ruptura de la diversificación: Durante un evento de riesgo de cola, activos que normalmente no tienen relación entre sí caen al mismo tiempo porque los inversores venden todo para conseguir liquidez.

**Diferencia con el "Cisne Negro"**

* Riesgo de cola: Es un evento extremo cuya probabilidad es muy baja, pero que se puede estimar estadísticamente observando el historial de crisis pasadas.
* Cisne Negro: Popularizado por Nassim Taleb, es un suceso completamente impredecible y sin precedentes históricos que solo se explica una vez ocurrido.

**Cómo se mide y se gestiona**

* Indicadores específicos. El ratio de Sharpe tradicional no captura bien este peligro porque asume una volatilidad simétrica. Para medir el riesgo de cola se utilizan métricas como el VaR (Valor en Riesgo) o el CVaR (Valor en Riesgo Condicional), que calculan la pérdida media esperada en el peor 1 % o 5 % de los escenarios.
* Cobertura de cartera. Los gestores protegen sus carteras comprando opciones financieras de venta (puts muy fuera de dinero), manteniendo liquidez o invirtiendo en activos refugio que suben cuando el pánico dispara la volatilidad.


### RATIO DE SHARPE

Creado por el premio Nobel William F. Sharpe, este indicador calcula el exceso de rentabilidad obtenido por cada unidad de volatilidad total.

El ratio de Sharpe y el ratio de Sortino son dos métricas financieras que miden la rentabilidad de una inversión en relación con el riesgo asumido. Ambos ayudan a determinar si las ganancias de una cartera provienen de una buena gestión o simplemente de asumir un riesgo excesivo.

* Fórmula y cálculo: Se resta la tasa libre de riesgo (como la rentabilidad de un bono estatal) al retorno de la inversión, y el resultado se divide entre la desviación estándar total de la cartera.

* Interpretación: Un valor superior a 1 se considera bueno, por encima de 2 es muy bueno y superior a 3 es excelente. Un valor negativo indica que la inversión rindió menos que un activo sin riesgo.

* Limitación principal: Penaliza por igual tanto las caídas bruscas como las subidas repentinas, ya que trata toda la volatilidad como algo negativo.

### RATIO DE SORTINO

Desarrollado por Frank Sortino, es una variante del anterior que mide el exceso de rentabilidad frente al riesgo a la baja (downside risk).

* Fórmula y cálculo. Mantiene el mismo numerador (rentabilidad obtenida menos tasa libre de riesgo o retorno mínimo aceptable), pero divide únicamente entre la desviación estándar de los rendimientos negativos.

* Interpretación. Al ignorar las oscilaciones al alza, suele arrojar cifras más altas que el de Sharpe (habitualmente entre 1,5 y 2 veces superiores en estrategias alcistas). Cuanto mayor sea el número, mejor protección ofrece ante pérdidas.

* Ventaja principal. Refleja con mayor precisión el riesgo real que preocupa al inversor, evitando castigar a los activos que tienen fuertes subidas de valor.

### NÚCLEO DE PRESERVACIÓN DE CAPITAL

Es una estrategia de inversión híbrida (tipo Core-Satellite o núcleo-satélite) diseñada para proteger el patrimonio en momentos de crisis y, al mismo tiempo, capturar las grandes subidas del mercado bursátil sin asumir el riesgo total de la bolsa.
El funcionamiento de esta arquitectura se divide en dos bloques con funciones opuestas pero complementarias:

1. El núcleo de preservación de capital (Defensa)

Representa la mayor parte de la cartera (habitualmente entre el 70 % y el 85 %) y su objetivo principal es evitar pérdidas graves y proteger el poder adquisitivo.
* Renta fija a corto plazo: Incluye letras del Tesoro o bonos gubernamentales de vencimiento muy cercano (menos de 1 a 3 años). Aportan liquidez inmediata, pagan intereses estables y apenas sufren caídas cuando suben los tipos de interés.
* Oro: Actúa como seguro frente a la inflación, la pérdida de valor de las divisas y las crisis geopolíticas. Suele comportarse bien cuando los activos financieros tradicionales entran en pánico.

2. La exposición controlada al momentum (Ataque)

Es la parte minoritaria de la cartera (entre el 15 % y el 30 %) y se encarga de generar crecimiento real por encima de la inflación.
* Qué es el momentum (impulso): Es una estrategia que consiste en comprar las acciones, sectores o índices que mejor se están comportando en los últimos meses, apostando a que esa tendencia alcista continuará durante un tiempo.
* Por qué es "controlada": No se mantiene invertida en bolsa de forma ciega. Utiliza reglas estrictas —como medias móviles o filtros de tendencia— para reducir la exposición o salir al efectivo en cuanto el mercado pierde fuerza o entra en tendencia bajista.

**Cómo se conecta con los conceptos anteriores**

Esta combinación está pensada específicamente para optimizar las métricas que se mencionan anteriormente:
* Mejora el ratio de Sortino: Al cortar las pérdidas rápidamente en la parte de renta variable y tener un núcleo muy estable, la volatilidad negativa es mínima.
* Reduce el riesgo de cola izquierda: Si la bolsa se desploma un 40 %, el impacto en el total de la cartera queda muy amortiguado porque la mayor parte del dinero está en bonos cortos y oro, y el momentum suele activar su salida antes de lo peor de la caída.
* Mantiene la cola derecha: Aunque solo tenga un 20 % o 30 % en acciones, al concentrarse en los activos con mayor impulso alcista, logra capturar buena parte de las grandes subidas del mercado.

Banco de España y Tesoro Público (datos.gob.es / INE): Utilizados como referencia macroeconómica para fijar la tasa libre de riesgo anualizada (aproximada al 3,25 % según el rendimiento medio de las Letras del Tesoro a 3 y 6 meses) y el umbral de inflación interanual (IPC), permitiendo evaluar si el rendimiento a corto plazo genera riqueza real.
